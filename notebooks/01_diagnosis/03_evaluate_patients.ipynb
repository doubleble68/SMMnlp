## Combine true and result data for performance


```python
final = true_testing.merge(
    results_all[["PatientSSN", "Note_index", "SMMDx_LLM_Binary"]],
    on=["PatientSSN", "Note_index"],
    how="inner",
)
```

## POST PROCESSING


```python
llm_columns = ["SMMDx_LLM_Binary", "SMM Dx (final)"]
final = add_parsed_labels(final, llm_columns, parse_diagnosis_label)
```

```python
final.head(5)
```

## Patient-level BINARY CLASSIFICATION PERFORMANCE


```python
final["post_SMM Dx (final)2"] = np.where(final["post_SMM Dx (final)"] == "Yes", 1, 0)
```

```python
final.head()
```

```python
ptlevel_true = final.groupby("PatientSSN")[["post_SMM Dx (final)2"]].sum().reset_index()
```

```python
ptlevel_true.head()
```

```python
ptlevel_true["post_SMM Dx (final)_Pt"] = np.where(
    ptlevel_true["post_SMM Dx (final)2"] >= 1, "Yes", "No"
)
ptlevel_true2 = ptlevel_true.drop(columns="post_SMM Dx (final)2")
```

```python
ptlevel_true2.head()
```

```python
len(ptlevel_true2)
```

```python
final_pos = final[final["post_SMMDx_LLM_Binary"] == "Yes"]
```

```python
df_sorted = final_pos.sort_values(["PatientSSN"])
final_pos2 = df_sorted.drop_duplicates(subset="PatientSSN")
```

```python
len(final_pos2)
```

```python
final_pos2.loc[:, "SMMDx_LLM_Binary_Pt"] = "Yes"
```

```python
final_pos = final_pos2[["PatientSSN", "SMMDx_LLM_Binary_Pt"]]
```

```python
final_pos.head(5)
```

```python
ptlevel_result = ptlevel_true2.merge(final_pos, on="PatientSSN", how="left")
ptlevel_result["SMMDx_LLM_Binary_Pt"] = ptlevel_result["SMMDx_LLM_Binary_Pt"].fillna("No")
```

```python
ptlevel_result.head(10)
```

```python
conditions = [
    (ptlevel_result["post_SMM Dx (final)_Pt"] == "Yes")
    & (ptlevel_result["SMMDx_LLM_Binary_Pt"] == "Yes"),
    (ptlevel_result["post_SMM Dx (final)_Pt"] == "No")
    & (ptlevel_result["SMMDx_LLM_Binary_Pt"] == "No"),
    (ptlevel_result["post_SMM Dx (final)_Pt"] == "Yes")
    & (ptlevel_result["SMMDx_LLM_Binary_Pt"] == "No"),
    (ptlevel_result["post_SMM Dx (final)_Pt"] == "No")
    & (ptlevel_result["SMMDx_LLM_Binary_Pt"] == "Yes"),
]
choices = ["TP", "TN", "FN", "FP"]
ptlevel_result["Outcome1_Binary_Pt"] = np.select(conditions, choices, default="Unknown")
```

```python
ptlevel_result["y_true"] = (
    ptlevel_result["post_SMM Dx (final)_Pt"].str.lower().map({"yes": 1, "no": 0})
)
ptlevel_result["y_pred"] = (
    ptlevel_result["SMMDx_LLM_Binary_Pt"].str.lower().map({"yes": 1, "no": 0})
)
```

```python
ptlevel_result.head(10)
```

```python
y_true = ptlevel_result["y_true"]
y_pred = ptlevel_result["y_pred"]
metric_result = calculate_binary_metrics(y_true, y_pred)
```

```python
cm_display = ConfusionMatrixDisplay(
    confusion_matrix=metric_result.cm, display_labels=["Non-SMM", "SMM"]
)
cm_display.plot()
plt.show()
```

```python
print_metrics(metric_result)
```
